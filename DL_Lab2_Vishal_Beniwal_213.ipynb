```python
import numpy as np

X = np.array([0.75, 0.65, 0.80,0.5,0.90])
w = np.array([0.4, 0.3, 0.5,0.3,0.6])
b = -0.2
y = 1
lr = 0.01

z = np.dot(X, w) + b
y_pred = 1 if z >= 0 else 0
error = y - y_pred

w = w + lr * error * X
b = b + lr * error

print("Net Input:", z)
print("Predicted Output:", y_pred)
print("Error:", error)
print("Updated Weights:", w)
print("Updated Bias:", b)
```
